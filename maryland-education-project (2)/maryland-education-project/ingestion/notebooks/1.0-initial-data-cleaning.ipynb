# 1.0 Initial Data Cleaning

DATA 110 Project 1: Maryland counties, education. This notebook reads the raw CSV, cleans it, and saves the result to `data/processed/`. Run the cells from top to bottom.

## 1. Setup

```python
import os
import pandas as pd

RAW = "../../data/raw/Choose_Maryland___Compare_Counties_Education.csv"
OUT = "../../data/processed/md_education_clean.csv"
URL = "https://raw.githubusercontent.com/A1-3x/DMV/main/compare_counties/Choose_Maryland___Compare_Counties_Education.csv"
```

## 2. Load the raw data
The raw file is never modified. If the repo copy is not found (for example in Colab), it is read from GitHub instead.

```python
raw = pd.read_csv(RAW if os.path.exists(RAW) else URL)
raw.head()
```

## 3. Clean the data
Drop the footer rows (Source, Date of Data, Note), use short column names, and turn the columns into numbers.

```python
# The file ends with footer rows that are not counties
df = raw[~raw["County"].isin(["Source", "Date of Data", "Note", "Quote"])].copy()

# Short column names
df.columns = ["county", "grads", "ratio", "spend", "ba", "hs", "cc2", "cc4", "enr2", "enr4"]
for c in df.columns[1:]:
    df[c] = pd.to_numeric(df[c])
df["short"] = df["county"].str.replace(" County", "", regex=False)

print("Rows:", len(df), "(23 counties + Baltimore City)")
df.head()
```

## 4. Check for missing values
Blank enrollment means the county has no college of that type.

```python
print("Missing values:\n", df.isna().sum().to_string())
```

## 5. Save the cleaned data

```python
os.makedirs(os.path.dirname(OUT), exist_ok=True)
df.to_csv(OUT, index=False)
print("Saved", OUT)
```
